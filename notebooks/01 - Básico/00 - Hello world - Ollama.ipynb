```python
import dspy  # Framework DSPy para trabalhar com Language Models
```

## Setup - Configuração do Modelo Local (Ollama)

Configuramos o DSPy para usar um modelo local (Qwen 3.5) rodando no Ollama. Diferente da API remota, Ollama é executado localmente em `http://localhost:11434` e não requer autenticação.

```python
lm = dspy.LM(
    "ollama_chat/qwen3.5:4b",  # Modelo local Qwen rodando no Ollama
    api_base="http://localhost:11434",  # URL do servidor Ollama local
    api_key="",  # Sem API key pois é local
)

# Configura o modelo padrão para todas as operações DSPy
dspy.configure(lm=lm)
```

```python
response = lm(
    "Explique em uma frase o que é uma árvore binária."
)  # Chamada ao modelo local Ollama

print(response)  # Imprime resposta completa (lista de dicts)
```

```python
print(response[0]['text'])  # Extrai apenas campo 'text' da resposta
```

```python
print(response[0].get('reasoning_content', 'Campo não disponível'))  # Tenta acessar raciocínio se existir
```

## Explorando Campos da Resposta

Alguns modelos (especialmente os com capacidade de chain-of-thought) retornam campos adicionais como `reasoning_content` com o raciocínio intermediário.

```python
# Mostra última chamada ao modelo (n=1 significa 1 última chamada)
dspy.inspect_history(n=1)
```

## Inspecionando o Histórico

DSPy mantém histórico de todas as chamadas ao modelo. Podemos inspecionar para debug, entender prompts gerados, custos, latência, etc.

### Acessando Histórico Programaticamente

Acesso direto ao histórico via `lm.history` para análise programática. Cada entrada contém prompt, resposta, metadados, etc.

```python
ultimo = lm.history[-1]  # Acessa última chamada no histórico
print(ultimo.keys())  # Mostra campos disponíveis (prompt, response, etc)
```

## Inspecionar Cada Campo

Itera sobre todos os campos da última chamada para visualizar prompt, resposta, latência, tokens, etc.

```python
for k in ultimo.keys():
    # Printa cada campo em formato key=value para fácil visualização
    print(f"{k=}, {ultimo[k]=}")
    print("-"*20)  # Separador para melhor legibilidade
```
