```python
import os
from dotenv import load_dotenv  # Carrega variáveis de ambiente do arquivo .env
import dspy  # Framework para otimização de prompts com Language Models

load_dotenv()  # Lê variáveis do arquivo .env
```

## Setup - Configuração do Modelo

Carregamos as variáveis de ambiente do arquivo `.env` na raiz do projeto. Isso evita hardcoding de credenciais no código.

```python
lm = dspy.LM(
    "openai/gpt-5-mini",  # Modelo OpenAI GPT-5 Mini
    api_key=os.getenv("OPENAI_API_KEY"),  # API key carregada da variável de ambiente
)

# Configura o modelo padrão para todas as operações DSPy
dspy.configure(lm=lm)
```

## Signature como Classe

Diferente da sintaxe em string, podemos definir Signatures como classes Python. Isso oferece:
- **Docstring**: instrução geral para o modelo
- **InputFields**: campos de entrada com descrições opcionais
- **OutputFields**: campos de saída com descrições opcionais

Essa abordagem é mais estruturada e clara para tarefas complexas.

```python
class PoemaNewton(dspy.Signature):
    """Escreva um poema respondendo à pergunta sobre um cientista famoso."""

    # Campo de entrada: nome do cientista
    cientista: str = dspy.InputField(desc="Nome do cientista sobre o qual perguntar")

    # Campo de entrada: aspecto a ser respondido
    aspecto: str = dspy.InputField(desc="O que se quer saber sobre o cientista")

    # Campo de saída: poema como resposta
    poema: str = dspy.OutputField(desc="Um poema que responde a pergunta sobre o cientista")
```

## Criar Preditor

Instancia um preditor com a Signature classe definida. O DSPy usa essa classe para estruturar prompts e parser respostas.

```python
# Cria um preditor que usa a Signature classe
gerador_poema = dspy.Predict(PoemaNewton)
```

## Executar Predição

Passamos valores para os InputFields. O modelo gera resposta estruturada no OutputField.

```python
resultado = gerador_poema(
    cientista="Isaac Newton",  # InputField: nome do cientista
    aspecto="local e ano de nascimento"  # InputField: aspecto perguntado
)

resultado  # Retorna Prediction object com campo .poema
```

## Acessar o Poema

Extrai o campo `poema` do objeto Prediction retornado.

```python
print(resultado.poema)  # Acessa o campo .poema com a resposta gerada
```

## Inspecionando o Histórico

```python
# Mostra última chamada ao modelo (n=1 significa 1 última chamada)
dspy.inspect_history(n=1)
```
