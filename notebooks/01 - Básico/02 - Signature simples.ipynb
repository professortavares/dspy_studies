````python
import os
from dotenv import load_dotenv  # Carrega variáveis de ambiente do arquivo .env
import dspy  # Framework para otimização de prompts com Language Models

load_dotenv()  # Lê variáveis do arquivo .env
````

## Setup - Configuração do Modelo

Carregamos as variáveis de ambiente do arquivo `.env` na raiz do projeto. Isso evita hardcoding de credenciais no código.

````python
lm = dspy.LM(
    "openai/gpt-5-mini",  # Modelo OpenAI GPT-5 Mini
    api_key=os.getenv("OPENAI_API_KEY"),  # API key carregada da variável de ambiente
)

# Configura o modelo padrão para todas as operações DSPy
dspy.configure(lm=lm)
````

## Signature em String

DSPy permite definir Signatures de forma simplificada usando strings. O formato é:
```
entrada1, entrada2 -> saida1, saida2
```

No nosso caso: `pergunta -> poema` (uma entrada, uma saída).

````python
# Define a Signature em formato de string simplificado
poema_signature = "pergunta -> poema"

# Cria um preditor com essa Signature
gerador_poema = dspy.Predict(poema_signature)
````

## Executar Predição

Passamos a pergunta como entrada e o modelo gera um poema como resposta.

````python
resultado = gerador_poema(
    pergunta="Qual é a terceira lei de Newton?"  # Pergunta a ser respondida em forma de poema
)

resultado  # Retorna Prediction object com campo .poema
````

## Acessar o Poema

Extrai o campo `poema` do objeto Prediction retornado.

````python
print(resultado.poema)  # Acessa o campo .poema com a resposta gerada
````

## Prompt criado

````python
# Mostra última chamada ao modelo (n=1 significa 1 última chamada)
dspy.inspect_history(n=1)
````

## Inspecionando o Histórico

DSPy mantém histórico de todas as chamadas ao modelo. Podemos inspecionar para debug, entender prompts gerados, custos, latência, etc.

````python
ultimo = lm.history[-1]  # Acessa última chamada no histórico
print(ultimo.keys())  # Mostra campos disponíveis (prompt, response, tokens, etc)
````

````python
for k in ultimo.keys():
    # Printa cada campo em formato key=value para fácil visualização
    print(f"{k=}, {ultimo[k]=}")
    print("-"*20)  # Separador para melhor legibilidade
````
