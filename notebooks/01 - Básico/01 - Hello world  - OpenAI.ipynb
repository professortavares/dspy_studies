```python
import os
from dotenv import load_dotenv  # Carrega variáveis de ambiente do arquivo .env
import dspy  # Framework para otimização de prompts com Language Models
```

```python
load_dotenv()  # Lê variáveis do arquivo .env
```

## Setup - Configuração do Modelo

Primeiro, carregamos as variáveis de ambiente (como API key) do arquivo `.env` na raiz do projeto. Isso evita hardcoding de credenciais no código.

```python
lm = dspy.LM(
    "openai/gpt-5-mini",  # Modelo OpenAI GPT-5 Mini (modelo compacto e rápido)
    api_key=os.getenv("OPENAI_API_KEY"),  # API key carregada da variável de ambiente
)

# Configura o modelo padrão para todas as operações DSPy
dspy.configure(lm=lm)
```

```python
response = lm(
    "Explique em uma frase quem foi Albert Einstein."
)  # Chamada ao modelo OpenAI

print(response)  # Imprime resposta completa (lista de dicts)
```

## Primeira Chamada Simples

Fazemos uma chamada direta ao modelo sem estrutura. O LM retorna uma lista com dicionários contendo texto, reasoning e outras informações.

```python
print(response[0])  # Acessa primeiro elemento: contém texto, finish_reason, etc
```

```python
# Mostra última chamada ao modelo (n=1 significa 1 última chamada)
dspy.inspect_history(n=1)
```

## Inspecionando o Histórico

DSPy mantém histórico de todas as chamadas ao modelo. Podemos inspecionar para debug, entender prompts gerados, custos, latência, etc.

### Acessando Histórico Programaticamente

Acesso direto ao histórico via `lm.history` para análise programática. Cada entrada contém prompt, resposta, metadados, etc.

```python
ultimo = lm.history[-1]  # Acessa última chamada no histórico
print(ultimo.keys())  # Mostra campos disponíveis (prompt, response, tokens, etc)
```

## Inspecionar Cada Campo

Itera sobre todos os campos da última chamada para visualizar prompt, resposta, latência, tokens, custos (OpenAI), etc.

```python
for k in ultimo.keys():
    # Printa cada campo em formato key=value para fácil visualização
    print(f"{k=}, {ultimo[k]=}")
    print("-"*20)  # Separador para melhor legibilidade
```
